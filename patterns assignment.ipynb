```python
from math import comb


def grid_row(cols, size):
    """Row of width `size` with '*' at the given columns."""
    cells = [" "] * size
    for c in cols:
        cells[c] = "*"
    return " ".join(cells).rstrip()


def full_row(size):
    return grid_row(range(0, size, 2), size)


n = 4
```

```python
## 1. Square Hollow Pattern
```

```python
def square_hollow(n):
    for i in range(n):
        if i in (0, n - 1):
            print("* " * n)
        else:
            print("*" + "  " * (n - 2) + " *")


square_hollow(n)
```

```python
## 2. Number Triangular
```

```python
def number_triangular(n):
    for i in range(1, n + 1):
        print(" " * (n - i) + " ".join(str(i) * i))


number_triangular(n)
```

```python
## 3. Number Increasing Pyramid
```

```python
def number_increasing_pyramid(n):
    for i in range(1, n + 1):
        print(*range(1, i + 1))


number_increasing_pyramid(n)
```

```python
## 4. Number Increasing Reverse Pyramid
```

```python
def number_increasing_reverse_pyramid(n):
    for i in range(n, 0, -1):
        print(*range(1, i + 1))


number_increasing_reverse_pyramid(n)
```

```python
## 5. Number Changing Pyramid
```

```python
def number_changing_pyramid(n):
    num = 1
    for i in range(1, n + 1):
        print(*range(num, num + i))
        num += i


number_changing_pyramid(n)
```

```python
## 6. Zero-One Triangle
```

```python
def zero_one_triangle(n):
    for i in range(1, n + 1):
        print(*[1 if (i + j) % 2 == 0 else 0 for j in range(1, i + 1)])


zero_one_triangle(n)
```

```python
## 7. Palindrome Triangular
```

```python
def palindrome_triangular(n):
    for i in range(1, n + 1):
        left = list(range(i, 0, -1))
        right = list(range(2, i + 1))
        print("  " * (n - i) + " ".join(map(str, left + right)))


palindrome_triangular(n)
```

```python
## 8. Rhombus Pattern
```

```python
def rhombus(n):
    for i in range(n):
        print(" " * (n - 1 - i) + "* " * n)


rhombus(n)
```

```python
## 9. Diamond Pattern
```

```python
def diamond(n):
    for i in range(1, n + 1):
        print(" " * (n - i) + "* " * i)
    for i in range(n - 1, 0, -1):
        print(" " * (n - i) + "* " * i)


diamond(n)
```

```python
## 10. Butterfly Star Pattern
```

```python
def butterfly(n):
    for i in range(1, n + 1):
        print("* " * i + "  " * (n - i) * 2 + "* " * i)
    for i in range(n - 1, 0, -1):
        print("* " * i + "  " * (n - i) * 2 + "* " * i)


butterfly(n)
```

```python
## 11. Square Fill Pattern
```

```python
def square_fill(n):
    for _ in range(n):
        print("* " * n)


square_fill(n)
```

```python
def reverse_left_half_pyramid(n):
    for i in range(n, 0, -1):
        print("  " * (n - i) + "* " * i)


reverse_left_half_pyramid(n)
```

```python
## 16. K Pattern
```

```python
def k_pattern(n):
    for i in range(n, 0, -1):
        print("* " * i)
    for i in range(2, n + 1):
        print("* " * i)


k_pattern(n)
```

```python
def triangle_star(n):
    for i in range(1, n + 1):
        print(" " * (n - i) + "* " * i)


triangle_star(n)
```

```python
## 18. Reverse Number Triangle Pattern
```

```python
def reverse_number_triangle(n):
    for i in range(1, n + 1):
        print(" " * (i - 1) + " ".join(map(str, range(i, n + 1))))


reverse_number_triangle(n)
```

```python
## 19. Mirror Image Triangle Pattern
```

```python
def mirror_image_triangle(n):
    for i in list(range(1, n + 1)) + list(range(n - 1, 0, -1)):
        print(" " * (i - 1) + " ".join(map(str, range(i, n + 1))))


mirror_image_triangle(n)
```

```python
## 20. Hollow Triangle Pattern
```

```python
def hollow_triangle(n):
    size = 2 * n - 1
    for i in range(n - 1):
        print(grid_row([n - 1 - i, n - 1 + i], size))
    print(full_row(size))


hollow_triangle(n)
```

```python
## 21. Hollow Reverse Triangle Pattern
```

```python
def hollow_reverse_triangle(n):
    size = 2 * n - 1
    print(full_row(size))
    for i in range(1, n):
        print(grid_row([i, size - 1 - i], size))


hollow_reverse_triangle(n)
```

```python
## 22. Hollow Diamond Pyramid
```

```python
def hollow_diamond(n):
    size = 2 * n - 1
    for i in list(range(n)) + list(range(n - 2, -1, -1)):
        print(grid_row([n - 1 - i, n - 1 + i], size))


hollow_diamond(n)
```

```python
## 23. Hollow Hourglass Pattern
```

```python
def hollow_hourglass(n):
    size = 2 * n - 1
    print(full_row(size))
    for i in range(1, n):
        print(grid_row([i, size - 1 - i], size))
    for i in range(1, n - 1):
        print(grid_row([n - 1 - i, n - 1 + i], size))
    print(full_row(size))


hollow_hourglass(n)
```

```python
## 24. Pascal's Triangle
```

```python
def pascals_triangle(n):
    for i in range(n):
        print("  " * (n - i - 1) + "   ".join(str(comb(i, j)) for j in range(i + 1)))


pascals_triangle(n)
```

```python
## 25. Right Pascal's Triangle
```

```python
def right_pascals_triangle(n):
    for i in list(range(1, n + 1)) + list(range(n - 1, 0, -1)):
        print("* " * i)


right_pascals_triangle(n)
```
